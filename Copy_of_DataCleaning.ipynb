🔴 C1 — Two Records Contain Customer Crisis Content

```python
pythonCRISIS_PHRASES = [
    "want to die", "don't want to live", "life feels meaningless",
    "buy it for myself to drink", "want to drink it",
]

def is_crisis_record(record):
    for msg in record.get("messages", []):
        content = msg.get("content", "").lower()
        if any(phrase in content for phrase in CRISIS_PHRASES):
            return True
    return False
```

🔴 C2 — Three Records Contain Real PII (Full Name + Phone + Address)

```python
import re

# Chinese mobile numbers (11 digits starting with 1[3-9])
PHONE_RE = re.compile(r'\b1[3-9]\d{9}\b')
# Chinese name + address patterns (heuristic)
ADDRESS_KEYWORDS = ['省', '市', '区', '县', '路', '街', '号', '村', 'Township', 'County', 'Province', 'Street', 'Road No.']

def redact_pii(text):
    text = PHONE_RE.sub('[PHONE_REDACTED]', text)
    return text

def has_address(text):
    return sum(1 for kw in ADDRESS_KEYWORDS if kw in text) >= 2
```

🔴 C5 — Systematic "Imately" Translation Artifact (46 records)

```python
import re

def fix_imately(text):
    # "imately" as a standalone word → "approximately"
    text = re.sub(r'\bimately\b', 'approximately', text)
    # "imately once" → "approximately once"
    return text
```

```python
import json

file_path = "/content/system_output_clean_11_11.jsonl"

old_word = "Imately"
new_word = "approximately"

# نقرأ ونعدل
updated_lines = []

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        record = json.loads(line)

        for msg in record.get("messages", []):
            if "content" in msg:
                msg["content"] = msg["content"].replace(old_word, new_word)

        updated_lines.append(json.dumps(record, ensure_ascii=False))

# 🔥 نرجع نكتب فوق نفس الملف (overwrite)
with open(file_path, "w", encoding="utf-8") as f:
    for line in updated_lines:
        f.write(line + "\n")

print("Done! File updated in place.")
```

🟡 R1 — Remove Greeting Artifacts from Mid-Conversation Turns (~80 instances)

```python
GREETING_RE = re.compile(
    r'^(Hello[,.]?\s*)?(I\'?m? ?here(?: to (?:assist|help)(?: you)?)?)[\.,]?\s*'
    r'|(Hello[,.]?\s*yes[,;]?\s*)|(Hello[,.]?\s*no[,;]?\s*)',
    re.IGNORECASE
)

def clean_greeting(text, turn_index):
    if turn_index == 0:
        return text  # keep greeting on first assistant turn
    return GREETING_RE.sub('', text).strip().capitalize()
```

🟡 R2 — Address 42 Very Short Assistant Responses

```python
def flag_low_quality_response(record):
    flags = []
    msgs = record.get("messages", [])
    for i in range(1, len(msgs)):
        msg = msgs[i]
        if msg.get("role") != "assistant": continue
        prev = msgs[i-1].get("content","").lower()
        content = msg.get("content","").strip()
        word_count = len(content.split())
        is_substantive_q = any(w in prev for w in ['how','what','when','which','why','where'])
        if word_count < 15 and is_substantive_q:
            flags.append({'turn': i, 'user': prev[:80], 'assistant': content})
    return flags
```

```python
import json

file_path = "/content/system_output_clean_22_22.jsonl"

results = []

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        record = json.loads(line)

        messages = record.get("messages", [])

        user_msg = None
        assistant_msg = None

        # استخراج user و assistant
        for msg in messages:
            if msg.get("role") == "user":
                user_msg = msg.get("content", "")
            elif msg.get("role") == "assistant":
                assistant_msg = msg.get("content", "")

        # شرط: assistant أقل من 25 حرف
        if assistant_msg and len(assistant_msg) < 25:
            results.append({
                "user": user_msg,
                "assistant": assistant_msg
            })

# طباعة النتائج
for i, r in enumerate(results, 1):
    print(f"\n--- Example {i} ---")
    print("USER:")
    print(r["user"])
    print("ASSISTANT (<25 chars):")
    print(r["assistant"])
```

🟢 O3 — Tilde Formatting Cleanup

```python
import json

file_path = "/content/system_output_clean_11_11.jsonl"

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

updated_lines = []

for line in lines:
    record = json.loads(line)

    for msg in record.get("messages", []):
        if msg.get("role") == "assistant":
            content = msg.get("content", "")

            # إزالة ~ من نهاية النص فقط
            msg["content"] = content.rstrip("~").rstrip()

    updated_lines.append(json.dumps(record, ensure_ascii=False))

# 🔥 overwrite نفس الملف
with open(file_path, "w", encoding="utf-8") as f:
    for line in updated_lines:
        f.write(line + "\n")

print("Done! File updated in place.")
```
